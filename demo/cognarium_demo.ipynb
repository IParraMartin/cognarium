```python
import cognarium as cg
```

```python
subject = cg.load("openbmb/MiniCPM5-1B", device="auto")
subject.device
```

# Running an Experiment

We can run an artificial grammar learning experiment using `paradigms.ArtificialGrammar()`.

```python
results = cg.run(
    cg.paradigms.ArtificialGrammar(
        n_train=16,
        n_test=8,
        noise=0.2,
        test="probability",
        n_participants=20
    ),
    subject
)
```

```python
results.score()
```

```python
table = results.to_pandas()
table
```

```python
negative = cg.tasks.Mapping.from_function(range(1, 21), lambda x: 21 - x, scale=range(1, 21))
paradigm = cg.paradigms.IteratedLearning(negative, generations=13, chains=2, n_prev_examples=10)
results = cg.run(paradigm, subject)
per_chain = results.metrics("correlation", reference=lambda x: x, by=["generation", "chain"])
per_chain.groupby("generation")["value"].median()
```
